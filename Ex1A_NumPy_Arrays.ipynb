# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[10, 11, 12],
                 [13, 14, 15]])

print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop works with entire rows of the array at once, so each row is multiplied by 2 together.
The second loop goes through each individual number inside the rows and multiplies each value separately.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# This result is different because x represents the whole row,
# so x * 2 multiplies the entire row each time instead of one value.
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA))

print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)

print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))

print(stacked_array_1)
```

```python
print(joined_array_1.shape)
print(joined_array_2.shape)

# Reshape joined_array_2 so it matches joined_array_1
joined_array_2 = joined_array_2.reshape(4, 3)

# Stack the arrays
stacked_array_2 = np.stack((joined_array_1, joined_array_2))

# Print result
print(stacked_array_2)
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))

print(another_joined_array)
```

```python
horizontal_stack = np.hstack((arrA, arrA))
# hstack joins arrays side by side horizontally

vertical_stack = np.vstack((arrA, arrA))
# vstack joins arrays vertically by rows

depth_stack = np.dstack((arrA, arrA))
# dstack joins arrays along a third depth dimension

print(horizontal_stack)
print(vertical_stack)
print(depth_stack)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)

print(splitsky)

# Result:
# 2 arrays are created.
# Each array is still two-dimensional.
```

```python
spliterino = np.array_split(joined_array_1, 3)

print(spliterino)

# Result:
# 3 arrays are created.
# Each array is still two-dimensional,
# but the arrays may not all contain the same number of rows.
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')

print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6],
                 [13, 5],
                 [7, 3]])

arrB_sorted = np.sort(arrB)

print(arrB_sorted)
```

```python
print(arrB)

z = arrB < 7

arrC = arrB[z]

print(arrC)
```

```python
y = arrB >= 7

arrD = arrB[y]

print(arrD)
```
