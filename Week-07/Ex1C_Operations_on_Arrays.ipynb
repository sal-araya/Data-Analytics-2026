```python
import numpy as np
```

## Sample Arrays

```python
# Random sample arrays

r_1 = np.random.rand(5)

r_2 = np.random.randint(1, 20, size=5)

r_3 = np.random.randn(4, 4)

r_4 = np.random.randint(1, 50, size=10)

r_5 = np.random.choice([10, 20, 30, 40, 50], size=6)

# Print arrays
print("r_1:\n", r_1)
print("\nr_2:\n", r_2)
print("\nr_3:\n", r_3)
print("\nr_4:\n", r_4)
print("\nr_5:\n", r_5)
```

## Permutations

```python
print("Original r_4:")
print(r_4)
```

```python
np.random.shuffle(r_4)

print("Shuffled r_4:")
print(r_4)
```

The elements in the array were rearranged into a random order.
The shuffle operation changed the original array directly instead of creating a new array.

```python
new_shuffle = np.random.permutation(r_4)

print("new_shuffle:")
print(new_shuffle)

print("\nr_4 after permutation:")
print(r_4)
```

`shuffle` changes the original array directly by rearranging its elements.

`permutation` creates a new shuffled version of the array while keeping the original array unchanged.

Use `shuffle` when you want to permanently randomize an array.
Use `permutation` when you want a shuffled copy without modifying the original data.

## Mathematical Operations

```python
print("r_1:")
print(r_1)
print("Shape of r_1:", r_1.shape)

print("\nr_2:")
print(r_2)
print("Shape of r_2:", r_2.shape)

result_1 = np.add(r_1, r_2)

print("\nr_1 + r_2:")
print(result_1)
print("Shape of result:", result_1.shape)
```

```python
print("r_1:")
print(r_1)
print("Shape of r_1:", r_1.shape)

print("\nr_3:")
print(r_3)
print("Shape of r_3:", r_3.shape)

try:
    result_2 = np.add(r_1, r_3)

    print("\nr_1 + r_3:")
    print(result_2)
    print("Shape of result:", result_2.shape)

except ValueError as e:
    print("\nError:")
    print(e)

# The arrays cannot be added because their shapes are incompatible for broadcasting.
```

```python
print("r_2:")
print(r_2)
print("Shape of r_2:", r_2.shape)

print("\nr_4:")
print(r_4)
print("Shape of r_4:", r_4.shape)

try:
    result_3 = np.add(r_2, r_4)

    print("\nr_2 + r_4:")
    print(result_3)
    print("Shape of result:", result_3.shape)

except ValueError as e:
    print("\nError:")
    print(e)

# The arrays cannot be added because their shapes are incompatible.
# Arrays with shapes (5,) and (10,) cannot be broadcast together.
```

```python
# r_1 + r_4
try:
    print("r_1 + r_4:")
    print(np.add(r_1, r_4))

except ValueError as e:
    print("Error for r_1 + r_4:")
    print(e)

print("\n------------------\n")

# r_2 + r_4
try:
    print("r_2 + r_4:")
    print(np.add(r_2, r_4))

except ValueError as e:
    print("Error for r_2 + r_4:")
    print(e)

print("\n------------------\n")

# r_3 + r_4
try:
    print("r_3 + r_4:")
    print(np.add(r_3, r_4))

except ValueError as e:
    print("Error for r_3 + r_4:")
    print(e)

# Findings:
# Arrays with incompatible shapes cannot be broadcast together.
# NumPy raises a ValueError when shapes do not match broadcasting rules
```

```python
print(r_4)

print("\nr_4 + np.array([1]):")
print(np.add(r_4, np.array([1])))
```

```python
print(r_4)

print("\nr_4 + np.array([[1]]):")
print(np.add(r_4, np.array([[1]])))
```

```python
print(r_4)

print("\nr_4 + np.array([1,1,1,1,1]):")

try:
    print(np.add(r_4, np.array([1,1,1,1,1])))

except ValueError as e:
    print("Error:")
    print(e)

# The arrays cannot be broadcast together because
# their shapes (10,) and (5,) are incompatible.
```

```python
print(r_4)

print("\nr_4 + np.array([[1],[1]]):")
print(np.add(r_4, np.array([[1],[1]])))

# Error finding:
# This produces a broadcasting error because the shapes are incompatible.
```

Broadcasting allows NumPy to perform operations on arrays with different shapes
when their dimensions are compatible.

When using the add function, NumPy matches elements by position. If one array
has fewer dimensions or contains a single value, NumPy automatically stretches
(or broadcasts) that value across the larger array.

The shape of the resulting array depends on the compatible dimensions of the
arrays being combined.

## Cumulative Operations

```python
print("r_4:")
print(r_4)
```

```python
print("Sum of all values in r_4:")
print(np.sum(r_4))
```

```python
print("Sum along axis 0:")
print(np.sum(r_4, axis=0))
```

```python
try:
    print("Sum along axis 1:")
    print(np.sum(r_4, axis=1))

except ValueError as e:
    print("Error:")
    print(e)

except Exception as e:
    print("Error:")
    print(e)

# r_4 is a one-dimensional array, so axis 1 does not exist.
```

## Grocery Problem

```python
# Categories:
# [vegetables, fruits, snacks, beverages]

bag1 = np.array([2, 3, 4, 0])

bag2 = np.array([4, 0, 1, 2])

print("bag1:")
print(bag1)

print("\nbag2:")
print(bag2)
```

```python
bag1_total = np.sum(bag1)
bag2_total = np.sum(bag2)

print(f"The first bag contains {bag1_total} items, and the second bag contains {bag2_total} items.")
```

```python
totals = np.add(bag1, bag2)

vegetables = totals[0]
fruits = totals[1]
snacks = totals[2]
beverages = totals[3]

print(f"There are {vegetables} vegetables, {fruits} fruits, {snacks} boxes of snacks, and {beverages} beverages in the two grocery bags.")
```

## Teamwork

```python
# Ali's door knocks from Monday to Friday
ali = np.array([5, 4, 6, 4, 2])

print("Ali:")
print(ali)
```

```python
# Ben missed Monday
ben = np.array([0, 5, 5, 3, 4])

print("Ben:")
print(ben)
```

```python
# Carla missed Wednesday
carla = np.array([3, 6, 0, 3, 4])

print("Carla:")
print(carla)
```

```python
all_knocks = np.array([ali, ben, carla])

print("All knocks:")
print(all_knocks)
```

```python
# Total for each person
ali_total = np.sum(ali)
ben_total = np.sum(ben)
carla_total = np.sum(carla)

# Group total
group_total = np.sum(all_knocks)

print(f"Ali knocked on {ali_total} doors.")
print(f"Ben knocked on {ben_total} doors.")
print(f"Carla knocked on {carla_total} doors.")

print(f"\nThe group collectively knocked on {group_total} doors.")
```

```python
# Total knocks per day
daily_totals = np.sum(all_knocks, axis=0)

print("Daily totals:")
print(daily_totals)

# Rolling total of knocks
rolling_knocks = np.cumsum(daily_totals)

print("\nRolling total of knocks:")
print(rolling_knocks)

# Estimated votes (1 vote for every 2 knocks)
estimated_votes = rolling_knocks * 0.5

print("\nRolling estimated votes:")
print(estimated_votes)
```
