```python
import numpy as np
import pandas as pd
```

## Series

```python
aa = [1, 2, 3, 4]

ab = (5, 6, 7, 8)

ac = {9, 10, 11, 12}

ad = {
    'first': 13,
    'second': 14,
    'third': 15,
    'fourth': 16
}
```

```python
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(list(ac))
ser4 = pd.Series(ad)

print(ser1)
print()

print(ser2)
print()

print(ser3)
print()

print(ser4)
```

## DataFrames from Series

```python
df1 = pd.DataFrame(ser1)

df1
```

```python
df1.insert(1, 'Second', ser2)
df1.insert(2, 'Third', ser3)

df1
```

```python
df1.insert(3, 'Fourth', ser4)

df1

# ser4 was created from a dictionary.
# The dictionary uses labels (keys) as indexes instead of numeric indexes.
# Because the indexes do not match df1's numeric indexes,
# pandas fills the rows with NaN values.
```

```python
df1.loc[4] = [5, 9, 12, 0]

df1
```

Some facts from the info() results:

Useful insights from describe():

Some summarizations may not be useful for ID columns because IDs are just identifiers and not meaningful measurements.
