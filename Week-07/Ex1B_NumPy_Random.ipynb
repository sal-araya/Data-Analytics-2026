```python
import numpy as np
```

## Generating Random Numbers

```python
r_1 = np.random.randint(1, 21, size=2)

print(r_1)
```

```python
r_2 = np.random.randint(1, 21, size=(1, 2))

print(r_2)
```

```python
r_3 = np.random.rand(2, 1)

print(r_3)
```

```python
r_4 = np.random.randint(20, 101, size=(4, 5))

print(r_4)
```

```python
r_5 = np.random.rand(4, 5)

print(r_5)
```

## Generating Random Choices from an Array

```python
students = np.array([
    "Nahom", "Daniel", "Lesley", "Jordan",
    "Naomi", "Danait", "Ahmad", "Lula",
    "Ruth", "Peniel", "Chris", "Dinu"
])

print(students)
```

```python
call_on = np.random.choice(students)

print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4))

print(breakout_groups)

# This is not the best way to assign breakout groups
# because some students may appear more than once
# while others may not appear at all.
```

```python
call_on_weighted = np.random.choice(
    students,
    size=5,
    p=[0.30, 0.10, 0.05, 0.10,
       0.05, 0.05, 0.10, 0.05,
       0.05, 0.05, 0.10, 0.00]
)

print(call_on_weighted)

# Probability explanation:
# John has a 30% chance of being selected.
# Mia has a 0% chance of being selected.
# Other students have smaller probabilitie
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))

print(bell_curve)
```
