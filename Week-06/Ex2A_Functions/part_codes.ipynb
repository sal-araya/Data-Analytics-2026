```python
## Lab 2. Parsing strings – part codes
```

```python
def get_supplier_code(part_code):
    for i in range(len(part_code)):
        if part_code[i] == ':':
            return part_code[:i]
```

```python
def get_product_num(part_code):
    colon_index = part_code.find(':')
    dash_index = part_code.find('-')

    return part_code[colon_index + 1:dash_index]
```

```python
def get_size(part_code):
    dash_index = part_code.find('-')

    return part_code[dash_index + 1:]
```

```python
part1 = 'ACME:123-L'
part2 = 'DI:12345-M'
part3 = 'ACE:1-12'
```

```python
supplier = get_supplier_code(part1)
product = get_product_num(part1)
size = get_size(part1)

print(f"For {part1}, the supplier is {supplier}.")
print(f"Part number {product} is available in size {size}.")
```

```python
supplier = get_supplier_code(part2)
product = get_product_num(part2)
size = get_size(part2)

print(f"For {part2}, the supplier is {supplier}.")
print(f"Part number {product} is available in size {size}.")
```

```python
supplier = get_supplier_code(part3)
product = get_product_num(part3)
size = get_size(part3)

print(f"For {part3}, the supplier is {supplier}.")
print(f"Part number {product} is available in size {size}.")
```
