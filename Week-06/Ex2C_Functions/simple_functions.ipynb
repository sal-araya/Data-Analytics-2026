```python
def favorite_things(name, movie, music):
    return f"""
Name: {name}
Favorite Movie: {movie}
Favorite Musician/Band: {music}
"""
```

```python
print(favorite_things(
    "Saliem Araya",
    "Black Panther",
    "The Weeknd"
))
```

```python
def why_im_here(purpose):
    return purpose
```

```python
print(why_im_here(
    "I joined the Year Up United Data Analyst program to build technical skills and prepare for a successful career in data analytics."
))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place(
    "New York City",
    "there is always something exciting to do and see"
))
```

```python
print(favorite_place(
    "Eritrea",
    "it reminds me of my culture, family, and childhood memories"
))
```
